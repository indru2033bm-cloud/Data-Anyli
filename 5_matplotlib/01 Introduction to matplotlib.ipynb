```python
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
runs = [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]

plt.plot(years,runs)
```

```python
plt.plot(years,runs)
plt.xlabel("Year")
plt.ylabel("Runs Scored")
plt.title("Sachin Tendulkur's Yearly Runs")
```

```python
kohli = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
sehwag = [0, 300, 800, 1200, 1500, 1700, 1600, 1400, 1000, 0]

plt.plot(years, kohli, label="Virat Kohli")
plt.plot(years, sehwag, label="Virender sehwag")
plt.plot(years, runs, label="Sachin Tendulker")

plt.xlabel("Year")
plt.ylabel("Runs Scored")
plt.title("Performance comparison")
plt.legend()
```

```python

plt.plot(years, kohli, 'ro-.', label="Virat Kohli")
plt.plot(years, sehwag, 'c*--', label="Virender sehwag")
plt.plot(years, runs, 'b^:',label="Sachin Tendulker")

plt.xlabel("Year")
plt.ylabel("Runs Scored")
plt.title("Performance comparison")
plt.legend()
```

```python
plt.plot(years, kohli, linewidth=3 ,label="Virat Kohli")
plt.plot(years, sehwag, linewidth=2, label="Virender sehwag")
plt.plot(years, runs,  linewidth = 2,label="Sachin Tendulker")

plt.xlabel("Year")
plt.ylabel("Runs Scored")
plt.title("Performance comparison")
plt.legend()
plt.grid(True)
```
