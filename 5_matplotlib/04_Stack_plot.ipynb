```python
import matplotlib.pyplot as plt
```

```python
days = [1 ,2 ,3, 4, 5, 6, 7]
studying = [3 ,2.5,2.5,3,2,3.5,3]
playing = [1,2,1,1,1,2,1]
watching_tv = [3, 4, 1,2, 1.5,0,2]
sleeping = [6.5,7,7,7,6.5,7,7]
```

```python
labels = ["Studing", "Playing","Watching TV" , "Sleeping"]
colors = ['skyblue', 'lightgreen', 'gold', 'lightcoral']
```

```python
plt.figure(figsize = (10,6))
plt.stackplot(days,studying,playing, watching_tv,sleeping,labels=labels, colors=colors,alpha=0.8)

plt.legend(loc = 'upper right')
plt.title("Weekly activity tracking")
plt.xlabel("Day")
plt.ylabel("Hours")
plt.grid(True)
```
